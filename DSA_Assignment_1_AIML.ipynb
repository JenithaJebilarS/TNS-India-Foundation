```python

# 1 if Statement

age = 18

if age >= 18:
    print("Eligible to vote")
```

```python
#2. if-else Statement

age = 16

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
#3. if-elif-else Statement

mark = 75

if mark >= 90:
    print("A Grade")
elif mark >= 60:
    print("B Grade")
else:
    print("C Grade")
```

```python
#4. Nested if
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
```

```python

#Loop Statements
#1 for loop
for i in range(1, 6):
    print(i)
```

```python
#2. while Loop
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# 3.Break
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

```python
# 4.Continue
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
#5.Pass
for i in range(1, 4):
    if i == 2:
        pass
    print(i)
```
