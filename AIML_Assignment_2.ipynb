```python
# 1.NumPy – Student Marks Analysis
import numpy as np

marks = np.array([85, 72, 90, 65, 78, 88, 55, 92, 70, 81])

print("Marks:", marks)

print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

print("Marks greater than 75:", marks[marks > 75])
```

```python
# 2. NumPy – Temperature Analysis
import numpy as np

temperature = np.array([28, 31, 33, 29, 35, 32, 30])

print("Temperature:", temperature)

print("Average Temperature:", np.mean(temperature))
print("Highest Temperature:", np.max(temperature))
print("Lowest Temperature:", np.min(temperature))

print("Temperature above 30°C:", temperature[temperature > 30])

updated_temperature = temperature + 2

print("Updated Temperature:", updated_temperature)
```

```python
#3.Pandas – Student Performance DataFrame
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Chitra", "Divya", "Ezhil", "Fathima", "Gokul", "Hari"],
    "Department": ["IT", "CSE", "IT", "ECE", "CSE", "IT", "ECE", "CSE"],
    "Marks": [85, 72, 90, 65, 78, 88, 55, 92],
    "Attendance": [90, 75, 95, 82, 78, 88, 70, 92]
}

df = pd.DataFrame(data)

print("First 5 Students:")
print(df.head())

print("\nAverage Marks:", df["Marks"].mean())

print("\nStudents scoring more than 75:")
print(df[df["Marks"] > 75])

print("\nStudents with attendance below 80%:")
print(df[df["Attendance"] < 80])

print("\nStudents sorted by marks:")
print(df.sort_values("Marks"))
```

```python
#4.Pandas – Product Sales Analysis
import pandas as pd

data = {
    "Product Name": ["Laptop", "Phone", "Headphones", "Keyboard", "Mouse"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories"],
    "Price": [50000, 25000, 2000, 1500, 800],
    "Quantity Sold": [10, 30, 60, 70, 100]
}

df = pd.DataFrame(data)

df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("Product Data:")
print(df)

print("\nProduct with highest sales:")
print(df.loc[df["Total Sales"].idxmax()])

print("\nAverage Product Price:", df["Price"].mean())

print("\nProducts with quantity sold greater than 50:")
print(df[df["Quantity Sold"] > 50])

print("\nProducts sorted by Total Sales:")
print(df.sort_values("Total Sales"))
```

```python
#5. Data Visualization – Monthly Sales
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]
sales = [20000, 25000, 22000, 30000, 28000, 35000]

# Line Chart
plt.plot(months, sales, marker="o", label="Sales")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()

# Bar Chart
plt.bar(months, sales, label="Sales")

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#6. Data Visualization – Student Marks
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
marks = [85, 72, 90, 65, 55, 88, 35, 78, 45, 92]

# Bar Chart
plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()


# Performance Categories
excellent = sum(80 <= m <= 100 for m in marks)
good = sum(60 <= m <= 79 for m in marks)
average = sum(40 <= m <= 59 for m in marks)
needs_improvement = sum(m < 40 for m in marks)

categories = ["Excellent", "Good", "Average", "Needs Improvement"]
values = [excellent, good, average, needs_improvement]

# Pie Chart
plt.pie(values, labels=categories, autopct="%1.1f%%")

plt.title("Student Performance Distribution")
plt.show()
```
